```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import drive
drive.mount('/content/drive')

# ============================================================
# 1. CONFIGURATION
# ============================================================

DATA_PATH = "/content/drive/MyDrive/COMP90055/adult-clean.csv"


# Change these to match your dataset
TARGET_ATTRIBUTE = "class_label"

PROTECTED_ATTRIBUTES = [
    "gender",
    "race",
    "age"
]

# Optional: attributes already known / suspected to be proxies
PROXY_ATTRIBUTES = [
    "relationship",
    "education",
    "occupation",
    "hours_per_week",
    "native_country"
]

OUTPUT_DIR = "eda_results"
os.makedirs(OUTPUT_DIR, exist_ok=True)


# ============================================================
# 2. LOAD DATA
# ============================================================
df = pd.read_csv(DATA_PATH)

# Normalise column names only
df.columns = [
    c.strip().lower().replace("-", "_")
    for c in df.columns
]

# Clean whitespace from string columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.strip()


print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
# ============================================================
# 3. BASIC DATASET STATISTICS
# ============================================================

summary = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing_count": df.isna().sum(),
    "missing_percent": df.isna().mean() * 100,
    "unique_values": df.nunique()
})

print("\n" + "=" * 60)
print("ATTRIBUTE SUMMARY")
print("=" * 60)

print(summary)

summary.to_csv(
    os.path.join(OUTPUT_DIR, "attribute_summary.csv")
)


print("\nNumber of duplicated rows:")
print(df.duplicated().sum())
```

```python
# ============================================================
# 4. NUMERIC ATTRIBUTE STATISTICS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns

numeric_stats = df[numeric_columns].describe().T

print("\n" + "=" * 60)
print("NUMERIC STATISTICS")
print("=" * 60)

print(numeric_stats)

numeric_stats.to_csv(
    os.path.join(OUTPUT_DIR, "numeric_statistics.csv")
)
```

```python
# ============================================================
# 5. CATEGORICAL ATTRIBUTE DISTRIBUTIONS
# ============================================================

categorical_columns = df.select_dtypes(
    include="object"
).columns

for column in categorical_columns:

    print("\n" + "=" * 60)
    print(f"DISTRIBUTION: {column}")
    print("=" * 60)

    counts = df[column].value_counts(dropna=False)

    percentages = (
        df[column]
        .value_counts(dropna=False, normalize=True)
        * 100
    )

    distribution = pd.DataFrame({
        "count": counts,
        "percentage": percentages
    })

    print(distribution)

    distribution.to_csv(
        os.path.join(
            OUTPUT_DIR,
            f"{column}_distribution.csv"
        )
    )
```

```python
# ============================================================
# 6. PROTECTED ATTRIBUTE DISTRIBUTIONS
# ============================================================

for protected in PROTECTED_ATTRIBUTES:

    if protected not in df.columns:
        print(f"{protected} not found.")
        continue

    print("\n" + "=" * 60)
    print(f"PROTECTED ATTRIBUTE: {protected}")
    print("=" * 60)

    if pd.api.types.is_numeric_dtype(df[protected]):

        print(df[protected].describe())

        plt.figure(figsize=(8, 5))
        plt.hist(
            df[protected].dropna(),
            bins=20
        )

        plt.xlabel(protected)
        plt.ylabel("Number of records")
        plt.title(
            f"Distribution of protected attribute: {protected}"
        )

    else:

        counts = df[protected].value_counts()

        percentages = (
            df[protected].value_counts(normalize=True)
            * 100
        )

        result = pd.DataFrame({
            "count": counts,
            "percentage": percentages
        })

        print(result)

        plt.figure(figsize=(8, 5))

        counts.plot(kind="bar")

        plt.xlabel(protected)
        plt.ylabel("Number of records")
        plt.title(
            f"Distribution of protected attribute: {protected}"
        )

        plt.xticks(rotation=45)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            f"protected_{protected}.png"
        )
    )

    plt.close()
```

```python
# ============================================================
# 7. TARGET DISTRIBUTION
# ============================================================

if TARGET_ATTRIBUTE in df.columns:

    target_counts = df[TARGET_ATTRIBUTE].value_counts()

    target_percentages = (
        df[TARGET_ATTRIBUTE]
        .value_counts(normalize=True)
        * 100
    )

    target_distribution = pd.DataFrame({
        "count": target_counts,
        "percentage": target_percentages
    })

    print("\n" + "=" * 60)
    print("TARGET DISTRIBUTION")
    print("=" * 60)

    print(target_distribution)

    target_distribution.to_csv(
        os.path.join(
            OUTPUT_DIR,
            "target_distribution.csv"
        )
    )

    plt.figure(figsize=(7, 5))

    target_counts.plot(kind="bar")

    plt.ylabel("Number of records")
    plt.title(
        f"Distribution of target: {TARGET_ATTRIBUTE}"
    )

    plt.xticks(rotation=0)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "target_distribution.png"
        )
    )

    plt.close()
```

```python
# ============================================================
# 8. TARGET DISTRIBUTION BY PROTECTED ATTRIBUTE
# ============================================================

for protected in PROTECTED_ATTRIBUTES:

    if protected not in df.columns:
        continue

    # For continuous age, handle separately below
    if pd.api.types.is_numeric_dtype(df[protected]):
        continue

    print("\n" + "=" * 60)
    print(
        f"{TARGET_ATTRIBUTE} BY {protected}"
    )
    print("=" * 60)

    counts = pd.crosstab(
        df[protected],
        df[TARGET_ATTRIBUTE]
    )

    proportions = pd.crosstab(
        df[protected],
        df[TARGET_ATTRIBUTE],
        normalize="index"
    )

    print("\nCounts:")
    print(counts)

    print("\nWithin-group proportions:")
    print(proportions)

    counts.to_csv(
        os.path.join(
            OUTPUT_DIR,
            f"{protected}_target_counts.csv"
        )
    )

    proportions.to_csv(
        os.path.join(
            OUTPUT_DIR,
            f"{protected}_target_proportions.csv"
        )
    )

    proportions.plot(
        kind="bar",
        figsize=(8, 5)
    )

    plt.ylabel("Proportion")
    plt.title(
        f"{TARGET_ATTRIBUTE} distribution by {protected}"
    )

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            f"{protected}_target_distribution.png"
        )
    )

    plt.close()
```

```python
import os
import pandas as pd

# ============================================================
# 1. CONFIGURATION
# ============================================================

OUTPUT_DIR = "proxy_analysis"

os.makedirs(OUTPUT_DIR, exist_ok=True)

TARGET_ATTRIBUTE = "class_label"

# Add the PADTAI proxy relations you want to inspect here.
#
# Example:
# relationship = husband  -> sex = male
#
# You can add more entries later.

PROXY_RELATIONSHIPS = [
    {
        "proxy_attribute": "relationship",
        "proxy_value": "Husband",
        "protected_attribute": "gender",
        "protected_value": "Male"
    },

    # Example placeholder:
    # {
    #     "proxy_attribute": "education",
    #     "proxy_value": "bachelors",
    #     "protected_attribute": "race",
    #     "protected_value": "white"
    # }
]


# ============================================================
# 2. LOAD DATA
# ============================================================


print("Dataset shape:", df.shape)
print(df.columns.tolist())


# ============================================================
# 3. ANALYSE PADTAI PROXY RELATIONSHIPS
# ============================================================

summary_rows = []

for relation in PROXY_RELATIONSHIPS:

    proxy_attr = relation["proxy_attribute"]
    proxy_value = relation["proxy_value"].lower()

    protected_attr = relation["protected_attribute"]
    protected_value = relation["protected_value"].lower()

    print("\n" + "=" * 70)
    print(
        f"{proxy_attr} = {proxy_value}"
        f"  ->  "
        f"{protected_attr} = {protected_value}"
    )
    print("=" * 70)

    # --------------------------------------------------------
    # Check columns exist
    # --------------------------------------------------------

    if proxy_attr not in df.columns:
        print(f"Column not found: {proxy_attr}")
        continue

    if protected_attr not in df.columns:
        print(f"Column not found: {protected_attr}")
        continue

    # --------------------------------------------------------
    # Filter rows containing the proxy value
    # --------------------------------------------------------

    proxy_rows = df[
        df[proxy_attr] == proxy_value
    ]

    proxy_count = len(proxy_rows)

    proxy_percentage = (
        proxy_count / len(df) * 100
    )

    print(f"\nProxy count: {proxy_count}")
    print(
        f"Proxy percentage of dataset: "
        f"{proxy_percentage:.2f}%"
    )


    # --------------------------------------------------------
    # Protected attribute distribution given proxy value
    #
    # P(protected | proxy)
    # --------------------------------------------------------

    protected_distribution = (
        proxy_rows[protected_attr]
        .value_counts(normalize=True)
        * 100
    )

    protected_counts = (
        proxy_rows[protected_attr]
        .value_counts()
    )

    protected_table = pd.DataFrame({
        "count": protected_counts,
        "percentage": protected_distribution
    })

    print(
        f"\nDistribution of {protected_attr} "
        f"given {proxy_attr} = {proxy_value}:"
    )

    print(protected_table)


    # --------------------------------------------------------
    # PADTAI-style proxy precision intuition
    #
    # P(protected_value | proxy_value)
    # --------------------------------------------------------

    if proxy_count > 0:

        matching_protected = (
            proxy_rows[protected_attr]
            == protected_value
        ).sum()

        conditional_probability = (
            matching_protected
            / proxy_count
        )

    else:

        matching_protected = 0
        conditional_probability = 0


    print(
        f"\nP({protected_attr}={protected_value}"
        f" | {proxy_attr}={proxy_value})"
    )

    print(
        f"{conditional_probability:.4f} "
        f"({conditional_probability * 100:.2f}%)"
    )


    # --------------------------------------------------------
    # Target distribution inside proxy group
    # --------------------------------------------------------

    if TARGET_ATTRIBUTE in df.columns:

        target_counts = (
            proxy_rows[TARGET_ATTRIBUTE]
            .value_counts()
        )

        target_percentages = (
            proxy_rows[TARGET_ATTRIBUTE]
            .value_counts(normalize=True)
            * 100
        )

        target_table = pd.DataFrame({
            "count": target_counts,
            "percentage": target_percentages
        })

        print(
            f"\nTarget distribution within "
            f"{proxy_attr}={proxy_value}:"
        )

        print(target_table)

        target_table.to_csv(
            os.path.join(
                OUTPUT_DIR,
                f"{proxy_attr}_{proxy_value}_target.csv"
            )
        )


    # --------------------------------------------------------
    # Save protected distribution
    # --------------------------------------------------------

    protected_table.to_csv(
        os.path.join(
            OUTPUT_DIR,
            f"{proxy_attr}_{proxy_value}_"
            f"{protected_attr}_distribution.csv"
        )
    )


    # --------------------------------------------------------
    # Add to summary table
    # --------------------------------------------------------

    summary_rows.append({
        "proxy_attribute": proxy_attr,
        "proxy_value": proxy_value,

        "protected_attribute": protected_attr,
        "protected_value": protected_value,

        "proxy_count": proxy_count,

        "proxy_percentage":
            proxy_percentage,

        "matching_protected_count":
            matching_protected,

        "P(protected|proxy)":
            conditional_probability
    })


# ============================================================
# 4. FINAL SUMMARY TABLE
# ============================================================

summary_df = pd.DataFrame(summary_rows)

print("\n" + "=" * 70)
print("PADTAI PROXY SUMMARY")
print("=" * 70)

print(summary_df)

summary_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "proxy_summary.csv"
    ),
    index=False
)
```
